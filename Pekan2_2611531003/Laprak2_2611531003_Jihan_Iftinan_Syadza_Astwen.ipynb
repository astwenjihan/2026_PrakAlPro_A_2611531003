## 2.1 Dasar Teori
1. Konsep Variabel: Wadah atau lokasi di memori komputer yang digunakan untuk menyimpan nilai data. Di Python, variabel tidak perlu dideklarasikan dengan tipe data tertentu secara eksplisit karena Python menganut sistem dynamically typed.
​
2. Tipe Data Numerik: Digunakan untuk menyimpan data angka. Terdiri dari Integer (int) untuk bilangan bulat, Float (float) untuk bilangan desimal, dan Complex (complex) untuk bilangan berjenis imajiner.

3. Tipe Data String (str): Digunakan untuk merepresentasikan teks atau kumpulan karakter yang diapit oleh tanda kutip tunggal, ganda, maupun tiga tanda kutip (""") untuk teks multi-baris.
4. Tipe Data Boolean (bool): Tipe data yang hanya memiliki dua nilai kebenaran mutlak, yaitu True (benar) dan False (salah), yang biasanya dihasilkan dari operasi perbandingan logika.

5. Input dari Pengguna (input()): Fungsi bawaan yang digunakan untuk menerima masukan data dari pengguna melalui terminal/konsol, di mana nilai yang diterima secara default bertipe data string dan dapat dikonversi (casting) ke bentuk lain.

6. Konstanta: Variabel yang nilainya bersifat tetap dan tidak boleh diubah sepanjang eksekusi program. Dalam konvensi bahasa Python, penamaan konstanta ditulis menggunakan huruf kapital seluruhnya.

##2.2.1 Numerik_2611531003.py

```python
#create a variable with integer value
a_1003=100
print("The type of variable having value", a_1003, "is", type(a_1003))

#create a variable with float value
b_1003=10.2345
print("The type of variable having value", b_1003, "is", type(b_1003))

#create a variable with complex value
c_1003=100+3j
print("The type of variable having value", c_1003, "is", type(c_1003))
```

**Penjelasan Kode:**

1. Variabel Integer (a_1003)
Kode: a_1003 = 100 lalu print(..., type(a_1003))
Penjelasan: Baris ini membuat variabel bernama a_1003 dan mengisinya dengan nilai bilangan bulat. Di Python, angka tanpa titik desimal otomatis dikategorikan sebagai tipe data int (integer). Fungsi type() akan mengembalikan keterangan <class 'int'>.


2. Variabel Float (b_1003)
Kode: b_1003 = 10.2345 lalu print(..., type(b_1003))
Penjelasan: Baris ini mendefinisikan variabel dengan nilai pecahan atau desimal. Nilai yang mengandung tanda titik akan otomatis terbaca oleh Python sebagai tipe data float (bilangan desimal). Fungsi type() mendeteksinya sebagai <class 'float'>.


3. Variabel Complex (c_1003)
Kode: c_1003 = 100 + 3j lalu print(..., type(c_1003))
Penjelasan: Baris ini mendeklarasikan bilangan kompleks yang terdiri dari bagian riil (100) dan bagian imajiner (3j, di mana huruf j mendefinisikan unit imajiner dalam matematika). Python mendukung tipe data ini secara langsung dan mengembalikannya sebagai <class 'complex'>.

4. Fungsi Utama yang Digunakan
type(): Fungsi bawaan Python (built-in function) untuk mendeteksi dan menampilkan jenis tipe data dari suatu variabel secara dinamis.
print(): Berfungsi untuk menampilkan teks, nilai variabel, serta hasil dari fungsi type() secara bersamaan ke layar terminal agar mudah dibaca.


**Penjelasan Output:**

1.  a_1003 = 100: Menyimpan data angka bulat (int).
​
2. b_1003 = 10.2345: Menyimpan data angka desimal (float).

3. ​c_1003 = 100 + 3j: Menyimpan dat
a angka imajiner/kompleks (complex).


4. type(): Fungsi bawaan Python untuk mengecek jenis tipe data variabel yang dicetak.

##2.2.3 Boolean_2611531003.py

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nama_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_1003 = True
is_cumlaude_1003 = True

# Menggunakan Boolean
nilai_1003 = 85
batas_lulus_1003 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_1003 = nilai_1003 >= batas_lulus_1003 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_1003)
print("Apakah lulus?", status_kelulusan_1003)
if is_lulus_1003 and is_cumlaude_1003:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan Kode:**
​

1. Tipe Data Boolean (is_lulus_1003, is_cumlaude_1003): Menyimpan nilai kebenaran murni, yaitu True (benar) atau False (salah).


2. Operator Perbandingan (>=): Membandingkan angka 85 dan 75 untuk menghasilkan nilai boolean secara otomatis ke dalam variabel status_kelulusan_1003.
​

3. Operator Logika (and): Menggabungkan dua kondisi boolean di dalam struktur percabangan if agar pesan cumlaude hanya tampil jika kedua variabel bernilai True.




**​Penjelasan Output:**


​Menampilkan judul pemeriksaan kelul
usan, mencetak nilai angka, menampilkan status kelulusan berupa nilai True, serta mengeksekusi blok if untuk mencetak kalimat "Selamat, Anda lulus dengan predikat Cumlaude!" ke layar karena seluruh syarat bernilai benar.

##2.2.4 Input_2611531003.py

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_1003 = input("Masukkan nilai panjang: ")
lebar_1003 = input("Masukkan nilai lebar: ")
luas_1003 = int(panjang_1003) * int(lebar_1003)
print("Luas =", luas_1003)
```

**Penjelasan Kode:**

1. ​Fungsi input() (panjang_1003, lebar_1003): Digunakan untuk menerima masukan data dari pengguna melalui terminal/konsol. Apapun yang diketik pengguna lewat fungsi ini akan otomatis dibaca sebagai tipe data string.

2. ​Konversi Tipe Data / Casting (int(...)): Mengubah nilai string dari input pengguna menjadi bilangan bulat (integer) agar bisa dioperasikan secara aritmatika, karena data teks murni tidak bisa dikalikan secara matematis.

3. Operasi Perkalian: Mengalikan variabel panjang dan lebar yang sudah dikonversi untuk mendapatkan hasil luas persegi panjang.

4. Fungsi print(): Berfungsi untuk mencetak teks keterangan beserta hasil akhir perhitungan ke layar terminal.


**Penjelasan Output:**

Program akan meminta pengguna memasukkan nilai panjang dan lebar secara berurutan. Jika pengguna memasukkan angka 10 untuk panjang dan 5 untuk lebar, program akan memproses perkaliannya dan mencetak teks Luas = 50 ke layar.

## 2.2.5 Konstanta_2611531003.py

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_1003 = float(input("Masukkan nilai jari-jari: "))
luas_1003 = PI * jari_1003 * jari_1003
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1003, luas_1003))
```

**Penjelasan Kode:**

1. ​from typing import Final & PI: Final = 3.14: Menggunakan modul Final untuk menandai variabel PI sebagai konstanta (nilai tetap) dengan penulisan huruf kapital sesuai konvensi umum pemrograman.

2. ​float(input(...)): Menerima masukan dari pengguna via terminal dan langsung mengonversinya menjadi bilangan desimal (float) agar siap dihitung secara matematis.

3. Perhitungan Luas (luas_1003 = PI * jari_1003 * jari_1003): Mengalikan nilai konstanta PI dengan kuadrat jari-jari dari input pengguna.

4.  Format String (%f & %.2f): Mengatur tampilan angka desimal agar tercetak rapi dengan batasan dua angka di belakang koma.


**​Penjelasan Output:**
Program akan menampilkan nilai PI, lalu meminta pengguna memasukkan angka jari-jari (contoh: 7). Setelah itu, program akan memproses rumus dan mencetak hasil akhirnya secara rapi, misalnya: “Luas lingkaran dengan jari-jari 7.00 adalah 153.86”.
