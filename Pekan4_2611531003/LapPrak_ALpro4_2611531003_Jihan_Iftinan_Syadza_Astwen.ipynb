## 2.1 Dasar Teori

Statement conditional (percabangan) atau kontrol alur (control flow) adalah mekanisme dalam pemrograman yang memungkinkan program untuk mengambil keputusan serta memilih jalur eksekusi kode tertentu berdasarkan hasil evaluasi nilai kebenaran (boolean: True atau False). Dalam bahasa pemrograman Python, struktur percabangan dikelompokkan berdasarkan jumlah kondisi dan alur keputusan yang ditangani:

1. Percabangan if Tunggal: Digunakan untuk mengevaluasi satu kondisi saja. Blok kode di dalam if hanya akan dieksekusi jika kondisi yang diuji bernilai True. Jika kondisi bernilai False, program langsung melompati blok tersebut dan melanjutkan eksekusi ke baris berikutnya.

2. Percabangan if-else: Digunakan untuk menangani dua kemungkinan jalur keputusan. Jika kondisi pada blok if bernilai True, maka perintah di dalam blok if dieksekusi. Namun jika bernilai False, alur program secara otomatis dialihkan untuk mengeksekusi perintah di dalam blok else.

3. Percabangan if-elif-else: Digunakan untuk mengevaluasi banyak kondisi (kondisi majemuk) secara berurutan. Program mengecek kondisi mulai dari if teratas hingga rentetan elif di bawahnya. Ketika ditemukan kondisi pertama yang bernilai True, blok kode tersebut dieksekusi dan seluruh pengecekan di bawahnya langsung dilewati. Blok else digunakan di akhir sebagai kondisi penampung (fallback).

4. Percabangan match-case: Merupakan fitur pencocokan struktur (pattern matching) yang efisien untuk menggantikan if-elif bertingkat pada pemilihan nilai diskrit. Nilai variabel dicocokkan secara presisi dengan skenario kasus (case 1, case 2, dst.), serta menggunakan simbol wildcard (case _) sebagai penangan default jika tidak ada kasus yang cocok.

5. Percabangan Multi-if Terpisah: Terdiri dari beberapa blok if independen yang dievaluasi secara mandiri satu per satu. Pendekatan ini memungkinkan akumulasi nilai atau penambahan kriteria bertingkat (seperti perhitungan akumulasi diskon) berjalan bersamaan tanpa saling menggagalkan.

6. Penggabungan Operator Logika pada Kondisi: Evaluasi syarat dalam struktur percabangan dapat digabungkan dengan operator logika and (semua syarat harus terpenuhi), or (salah satu syarat terpenuhi), dan not (pembalikan nilai kebenaran). Penggunaan operator logika ini memungkinkan penanganan kondisi kompleks yang tetap bersih dan terstruktur tanpa memerlukan percabangan bersarang (nested-IF).

## 2.2.1 If1_2611531003.py

```python
ipk_1003 = float(input("Input IPK Anda ="))

if ipk_1003 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK" + str(ipk_1003))

print("Program Selesai")
```

**Penjelasan Kode:**

1. Input dan Konversi Data (`ipk_1003 = float(input(...))`): Berfungsi menerima masukan IPK dari pengguna melalui terminal, lalu dikonversi menjadi tipe data desimal (`float`) agar dapat dibandingkan secara numerik.

2. Percabangan IF Tunggal (`if ipk_1003 > 2.75:`): Memeriksa apakah nilai IPK lebih besar dari 2.75. Jika bernilai `True`, program mengeksekusi perintah cetak kelulusan di dalamnya.

3. Pencetakan Akhir (`print("Program Selesai")`): Perintah di luar blok `if` yang akan selalu dijalankan di akhir program.
​

**Penjelasan Output:**

1. Status Kelulusan (Kondisi True): Menampilkan pesan kelulusan "Anda Lulus Sangat Memuaskan dengan IPK..." beserta nilai IPK jika pengguna memasukkan nilai IPK lebih besar dari 2.75 (> 2.75). Jika IPK $\le$ 2.75, baris ini dilewati.

2. Penanda Program Selesai: Menampilkan pesan "Program Selesai" di akhir layar sebagai penanda eksekusi telah berakhir, yang akan selalu muncul baik kondisi IPK memenuhi syarat maupun tidak.

## 2.2.2 If2_2611531003.py

```python
ipk_1003 = float(input("Input IPK Anda ="))

if ipk_1003 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK" + str(ipk_1003))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

**Penjelasan Kode:**

1. Input dan Konversi Data (ipk_1003 = float(input(...))): Berfungsi untuk menerima masukan nilai IPK dari pengguna melalui terminal secara interaktif, kemudian langsung dikonversi menjadi tipe data desimal (float) agar dapat dibandingkan secara numerik.

2. Percabangan IF-ELSE (if ... else:): Memeriksa kondisi logika apakah nilai ipk_1003 lebih besar dari 2.75. Jika bernilai True, program menjalankan blok if. Jika bernilai False (IPK $\le$ 2.75), program mengalihkan eksekusi ke blok else.

3. Blok IF & Penggabungan String (str(ipk_1003)): Mengeksekusi perintah cetak kelulusan dengan mengonversi nilai ipk_1003 menjadi teks (string) agar dapat digabungkan (concatenation) menggunakan operator +.

4. Blok ELSE: Berisi perintah print("Anda Tidak Lulus") yang khusus dieksekusi jika kondisi if tidak terpenuhi.

5. Pencetakan Luar Blok (print("Program Selesai")): Berada di luar struktur if-else (tanpa indentasi), sehingga perintah ini akan selalu dieksekusi di akhir program baik kondisi bernilai True maupun False.


**​Penjelasan Output:**

1. Status Kelulusan (Kondisi True): Menampilkan pesan kelulusan "Anda Lulus Sangat Memuaskan dengan IPK..." beserta nilai IPK jika pengguna memasukkan nilai IPK lebih besar dari 2.75 (> 2.75).

2. Status Tidak Lulus (Kondisi False): Menampilkan pesan "Anda Tidak Lulus" jika pengguna memasukkan nilai IPK kurang dari atau sama dengan 2.75 (<= 2.75).

3. Penanda Program Selesai: Menampilkan pesan "Program Selesai" di bagian akhir layar sebagai penanda eksekusi telah berakhir, yang akan selalu muncul terlepas dari hasil evaluasi kondisi if-else.

## 2.2.3 If_elif_else1_2611531003.py

```python
umur_1003 = int(input("Input umur anda: "))
sim_1003 = input("Apakah Anda Sudah Punya SIM C: ")[0]

if umur_1003 >= 17 and sim_1003 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_1003 >= 17 and sim_1003 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_1003 < 17 and sim_1003 == 'y':
    print("Anda belum cukup umur untuk punya SIM")
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan Kode:**

1. Input dan Konversi Data (umur_1003 & sim_1003): Berfungsi untuk menerima masukan dari pengguna melalui terminal secara interaktif. Nilai umur langsung dikonversi menjadi integer (int()), sedangkan variabel status SIM C hanya mengambil karakter pertama dari teks input menggunakan pengindeksan [0].

2. Struktur Percabangan IF-ELIF-ELSE: Menggunakan satu alur bercabang yang saling terhubung secara berurutan. Jika salah satu kondisi di atasnya sudah bernilai True, program langsung mengeksekusi blok tersebut dan langsung melompati seluruh cek kondisi elif maupun else di bawahnya.

3. Operator Logika AND (and) & Perbandingan: Mengombinasikan dua syarat sekaligus dalam setiap pengecekan kondisi, yaitu batasan usia 17 tahun dan kepemilikan SIM C ('y' atau selain 'y').

4. Blok Terakhir ELSE: Berfungsi sebagai kondisi penampung (fallback) yang dieksekusi secara otomatis jika seluruh kondisi if dan elif di atasnya bernilai False (yaitu ketika umur < 17 dan SIM != 'y').

5. Pencetakan Luar Blok (print("Program Selesai")): Berada di luar struktur percabangan (tanpa indentasi), sehingga perintah ini akan selalu dieksekusi di akhir program.


**​Penjelasan Output:**

1. Dewasa dan Punya SIM (Kondisi IF Pertama): Menampilkan pesan "Anda sudah dewasa dan boleh bawa motor" jika pengguna berumur 17 tahun ke atas ($\ge$ 17) dan menginputkan memiliki SIM (sim_1003 == 'y').

2. Dewasa Tanpa SIM (Kondisi ELIF Pertama): Menampilkan pesan "Anda sudah dewasa tetapi tidak boleh bawa motor" jika pengguna berumur 17 tahun ke atas ($\ge$ 17) dan tidak memiliki SIM (sim_1003 != 'y').

3. Belum Cukup Umur Punya SIM (Kondisi ELIF Kedua): Menampilkan pesan "Anda belum cukup umur untuk punya SIM" jika pengguna berumur di bawah 17 tahun ($<$ 17) dan menginputkan memiliki SIM (sim_1003 == 'y').

4. Belum Cukup Umur Tanpa SIM (Kondisi ELSE): Menampilkan pesan "Anda belum cukup umur dan tidak boleh bawa motor" jika semua kondisi sebelumnya tidak terpenuhi (umur $<$ 17 dan tidak memiliki SIM).

5. Penanda Program Selesai: Menampilkan pesan "Program Selesai" di bagian akhir layar sebagai penanda eksekusi telah berakhir, yang akan selalu muncul terlepas dari kondisi mana yang terpenuhi.

## 2.2.5 Multi_if2_2611531003.py

```python
# Input dari user
total_belanja_1003 = float(input("Masukkan total belanja (Rp): "))

# Input dari status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1003 = input("Apakah anda member (y/t): ").strip().lower()
is_member_1003 = input_member_1003 in ['y', 'ya']

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_1003 = input("Apakah kode promo valid? (y/t):").strip().lower()
kode_promo_valid_1003 = input_promo_1003 in["y", "ya"]

total_diskon_persen_1003 = 0

if total_belanja_1003 > 10000000:
    total_diskon_persen_1003 +=10 #Diskon belanja besar

if is_member_1003:
    total_diskon_persen_1003 += 5 #Diskon member

if kode_promo_valid_1003:
    total_diskon_persen_1003 +=15 #Diskon voucher

#Menghitung nominal diskon dan total bayar
nominal_diskon_1003 = total_belanja_1003 * (total_diskon_persen_1003 / 100)
total_bayar_1003 = total_belanja_1003 - nominal_diskon_1003

#Output hasil
print("/n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_1003}%(Rp{nominal_diskon_1003:,.0f})")
print(f"Total Bayar: Rp {total_bayar_1003:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_1003}%")
# Output total diskon yang anda dapatkan: 30% jika belanja > 1juta, member, dan kode promo valid
```

**Penjelasan Kode:**

1. Input dan Pemrosesan String (input(), .strip().lower()): Berfungsi untuk menerima masukan dari pengguna melalui terminal (total belanja, status member, dan keabsahan promo). Teks masukan dibersihkan dari spasi berlebih dengan .strip() dan diubah menjadi huruf kecil dengan .lower().

2. Pemeriksaan Keanggotaan (in ['y', 'ya']): Menilai apakah input pengguna bernilai boolean True atau False berdasarkan keberadaan teks di dalam list ['y', 'ya'].

3. Akumulasi Diskon Bertingkat (Multi-IF Terpisah): Menggunakan 3 blok if terpisah (independen) tanpa elif untuk menambahkan nilai persentase diskon secara kumulatif (+=) sesuai syarat yang terpenuhi.

4. Kalkulasi Nominal & Total Bayar: Menghitung besarnya nominal potongan dalam Rupiah (nominal_diskon_1003) serta menentukan sisa total pembayaran akhir (total_bayar_1003).

5. Pemformatan Output (f-string & :,.0f): Mencetak rincian transaksi dengan format desimal tanpa angka di belakang koma (:,.0f) dan pemisah ribuan agar mudah dibaca.


**​Penjelasan Output:**

1. Diskon Belanja Besar (Multi-IF 1): Menambahkan persentase diskon sebesar 10% jika pengguna menginputkan total belanja lebih dari Rp10.000.000 (> 10000000).

2. Diskon Status Member (Multi-IF 2): Menambahkan persentase diskon sebesar 5% jika variabel is_member_1003 bernilai True (pengguna menginput 'y' atau 'ya').

3. Diskon Kode Promo (Multi-IF 3): Menambahkan persentase diskon sebesar 15% jika variabel kode_promo_valid_1003 bernilai True (pengguna menginput 'y' atau 'ya').

4. Rincian Pembayaran: Menampilkan gabungan total persentase diskon beserta nominal potongannya dalam Rupiah, serta menampilkan Total Bayar akhir setelah dipotong diskon.

5. Ringkasan Total Diskon: Menampilkan rangkuman total persentase diskon yang berhasil didapatkan oleh pengguna dari akumulasi seluruh kondisi yang terpenuhi (maksimal hingga 30%).

## 2.2.5 Match_case1_2611531003.py

```python
bulan_1003 = int(input("Masukkan angka bulan (1-12): "))

match bulan_1003:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan Kode:**

1. Input dan Konversi Data (bulan_1003 = int(input(...))): Berfungsi untuk menerima masukan angka dari pengguna melalui terminal secara interaktif, kemudian langsung dikonversi menjadi tipe data integer (int) agar siap dicocokkan nilainya pada percabangan.

2. Struktur Percabangan Match-Case (match bulan_1003:): Mengimplementasikan structural pattern matching untuk mencocokkan nilai variabel bulan_1003 secara presisi ke dalam beberapa opsi kasus (case) tanpa harus menggunakan banyak urutan elif.

3. Pencocokan Kasus Spesifik (case 1: hingga case 12:): Menilai apakah masukan angka bernilai 1 sampai 12, lalu mengeksekusi perintah cetak nama bulan yang sesuai (Januari hingga Desember).

4. Kondisi Default / Wildcard (case _:): Berfungsi sebagai penampung (fallback) yang dieksekusi secara otomatis jika nilai bulan_1003 tidak cocok dengan angka 1 sampai 12.


**​Penjelasan Output:**

1. Nama Bulan Valid (Case 1 - 12): Menampilkan nama bulan yang sesuai dalam bahasa Indonesia (seperti "Januari" untuk input 1, "Februari" untuk input 2, hingga "Desember" untuk input 12) jika pengguna memasukkan angka bernilai 1 sampai 12.

2. Peringatan Angka Tidak Valid (Case Wildcard _): Menampilkan pesan "Angka tidak valid" jika pengguna memasukkan angka di luar rentang 1 sampai 12 (seperti angka 0, 13, atau angka negatif).

## 2.2.6 Nilai_2611531003.py

```python
nilai_1003 = int(input("inputkan nilai angka= "))

if nilai_1003 >=81:
    print("A")
elif nilai_1003 >=70:
    print("B")
elif nilai_1003 >=60:
    print("C")
elif nilai_1003 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan Kode:**

1. Input dan Konversi Data (nilai_1003 = int(input(...))): Berfungsi untuk menerima masukan nilai angka dari pengguna melalui terminal secara interaktif, kemudian langsung dikonversi menjadi tipe data integer (int) agar siap dibandingkan secara numerik.

2. Struktur Percabangan IF-ELIF-ELSE: Menggunakan alur percabangan bertingkat untuk mengevaluasi rentang nilai secara berurutan dari batas tertinggi ke batas terendah. Ketika salah satu kondisi bernilai True, program mengeksekusi blok tersebut dan langsung melompati pengecekan di bawahnya.

3. Pengecekan Kondisi Batas (>= 81, >= 70, >= 60, >= 50): Menentukan standar batas minimal angka untuk mengelompokkan kelulusan nilai ke dalam kategori huruf mutu A, B, C, atau D.

4. Blok Terakhir ELSE: Berfungsi sebagai penampung (fallback) yang dieksekusi secara otomatis jika masukan angka tidak memenuhi seluruh kriteria di atas (yaitu nilai kurang dari 50).

**​Penjelasan Output:**

1. Huruf Mutu A (Kondisi IF Pertama): Menampilkan huruf "A" jika pengguna memasukkan nilai angka lebih dari atau sama dengan 81 ($\ge 81$).

2. Huruf Mutu B (Kondisi ELIF Pertama): Menampilkan huruf "B" jika pengguna memasukkan nilai angka di rentang 70 sampai 80 ($70 \le \text{nilai} < 81$).

3. Huruf Mutu C (Kondisi ELIF Kedua): Menampilkan huruf "C" jika pengguna memasukkan nilai angka di rentang 60 sampai 69 ($60 \le \text{nilai} < 70$).

4. Huruf Mutu D (Kondisi ELIF Ketiga): Menampilkan huruf "D" jika pengguna memasukkan nilai angka di rentang 50 sampai 59 ($50 \le \text{nilai} < 60$).

5. Huruf Mutu E (Kondisi ELSE): Menampilkan huruf "E" jika pengguna memasukkan nilai angka di bawah 50 ($< 50$).

**Deskripsi Tugas:**
Membuat sistem loket berbasis  konsol untuk menentukan jenis wahana, memverifikasi kelayakan pengunjung dengan kondisi majemuk (logical operator), menghitung diskon akumulatif menggunakan multi-IF terpisah, serta mencetak audit status operasional.

**Ketentuan Teknis Program:**
1. **Aturan Penamaan Variabel**:
   - Wajib diakhiri dengan 4 digit terakhir NIM (contoh: `_1003`).
   - Bersifat *case-sensitive* dan tidak boleh menggunakan *reserved keywords* Python.

2. **IF Tunggal**: Validasi kelayakan `jumlah_tiket <= 0`.

3. **Match-Case**: Pemilihan paket wahana (1–5) dan default (`_`).

4. **IF-ELIF-ELSE + Logika (`and`, `!=`)**: Validasi izin kendali khusus Paket 3 (Motor ATV) dan wahana lainnya.

5. **Multi-IF Terpisah**: Akumulasi diskon bertingkat (Subtotal $\ge$ Rp200rb, Member, Promo, Rombongan).

6. **IF-ELSE**: Evaluasi bonus souvenir jika `total_bayar > 300000`.

**Jawaban:**

```python
# Program Studi Kasus: Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
# Nama File: tugas4_2611531003.py

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

#1 Input Data Pengunjung & String Handling
nama_pengunjung_1003 = input("Masukkan nama pengunjung: ")
umur_pengunjung_1003 = int(input("Input umur anda: "))
sim_input_1003 = input("Apakah anda memiliki SIM-C? (y/t): ")
sim_pengunjung_1003 = sim_input_1003.strip().lower()[0] if sim_input_1003 else 't'
jumlah_tiket_1003 = int(input("Masukkan jumlah tiket  : "))

# Penerapan IF Tunggal untuk validasi kuota tiket
if jumlah_tiket_1003 <= 0:
    print("\nPeringatan: Kuota tiket tidak valid!")

# Tampilan Menu Wahana
print("\nPilihan Paket Wahana (1-5):")
print("1. Safari Rimba         (Rp 50,000)")
print("2. Arung Jeram          (Rp 75,000)")
print("3. Motor ATV Ekstrim    (Rp 120,000)")
print("4. Roller Coaster Kilat (Rp 100,000)")
print("5. All-Access VIP       (Rp 220,000)")

paket_1003 = int(input("Masukkan nomor paket (1-5) : "))

# 2. Pemilihan Wahana Menggunakan match - case
is_valid_paket_1003 = True
nama_wahana_1003 = ""
harga_satuan_1003 = 0

match paket_1003:
    case 1:
        nama_wahana_1003 = "Wahana Safari Rimba"
        harga_satuan_1003 = 50000
    case 2:
        nama_wahana_1003 = "Wahana Arung Jeram"
        harga_satuan_1003 = 75000
    case 3:
        nama_wahana_1003 = "Wahana Motor ATV Ekstrim"
        harga_satuan_1003 = 120000
    case 4:
        nama_wahana_1003 = "Wahana Roller Coaster Kilat"
        harga_satuan_1003 = 100000
    case 5:
        nama_wahana_1003 = "Wahana All-Access VIP"
        harga_satuan_1003 = 220000
    case _:
        print("Paket wahana tidak valid!")
        is_valid_paket_1003 = False

# Memastikan transaksi hanya berlanjut jika paket & tiket valid
if is_valid_paket_1003 and jumlah_tiket_1003 > 0:
    member_input_1003 = input("Apakah Anda member? (y/t) : ")
    is_member_1003 = member_input_1003.strip().lower()[0] if member_input_1003 else 't'

    promo_input_1003 = input("Apakah kode promo valid? (y/t) : ")
    kode_promo_valid_1003 = promo_input_1003.strip().lower()[0] if promo_input_1003 else 't'

    # 3. Validasi Izin Kendali Wahana Menggunakan if - elif - else & Operator Logika
    print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
    if paket_1003 == 3:
        if umur_pengunjung_1003 >= 17 and sim_1003 == 'y':
            print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
        elif umur_pengunjung_1003 >= 17 and sim_1003 != 'y':
            print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_pengunjung_1003 < 17 and sim_1003 == 'y':
            print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
    else:
        if umur_pengunjung_1003 >= 10:
            print("Status Akses: Usia memenuhi syarat untuk wahana ini.")
        else:
            print("Status Akses: Pengunjung di bawah umur 10 tahun wajib dalam pengawasan orang tua.")

    # 4. Akumulasi Diskon Bertingkat Menggunakan Multi-IF Terpisah
    subtotal_1003 = harga_satuan_1003 * jumlah_tiket_1003
    total_diskon_persen_1003 = 0

    if subtotal_1003 >= 200000:
        total_diskon_persen_1003 += 10
    if is_member_1003 in ['y', 'ya']:
        total_diskon_persen_1003 += 5
    if kode_promo_valid_1003 in ['y', 'ya']:
        total_diskon_persen_1003 += 15
    if jumlah_tiket_1003 >= 5:
        total_diskon_persen_1003 += 5

    # 5. Evaluasi Kelulusan Audit Menggunakan if - else
    nominal_diskon_1003 = subtotal_1003 * (total_diskon_persen_1003 / 100)
    total_bayar_1003 = subtotal_1003 - nominal_diskon_1003

    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp{int(subtotal_1003):,}".replace(",", "."))
    print(f"Total Diskon     : {total_diskon_persen_1003}% (Rp{int(nominal_diskon_1003):,})".replace(",", "."))
    print(f"Total Bayar      : Rp{int(total_bayar_1003):,}".replace(",", "."))

    if total_bayar_1003 > 300000:
        print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
    else:
        print("Catatan Layanan  : Terima kasih telah berkunjung.")

    print("Program Selesai")
```

**Penjelasan:**

**Penjelasan Program** (`Tugas4_2611531003.py`)

Program ini dirancang untuk mendemonstrasikan penggunaan 4 jenis Statement Conditional pada Python melalui studi kasus **Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana**. Seluruh variabel utama menggunakan konvensi penamaan yang diakhiri dengan 4 digit NIM (`_1003`).

Berikut adalah penjelasan detail alur logika dan teknis dari program berdasarkan urutan penulisan kode:


1. **String Handling & Input**: Menggunakan `.strip().lower()[0]` untuk mengambil karakter pertama dari input user secara konsisten (`'y'` atau `'t'`).
2. **Validation IF Tunggal**: Memeriksa `jumlah_tiket_1003 <= 0` untuk mendeteksi input tiket yang tidak valid.
3. **Match-Case Structure**: Mengatur variabel `nama_wahana_1003` dan `harga_satuan_1003` berdasarkan nomor paket. Menggunakan wildcard `_` untuk penanganan kesalahan input paket.
4. **Validasi Kualifikasi Pengendara**: Mengevaluasi syarat usia dan SIM C khusus Paket 3 menggunakan gabungan operator logika (`and`, `!=`).
5. **Multi-IF Akumulasi Diskon**: 4 blok `if` terpisah berjalan secara independen untuk menambahkan persen diskon tanpa adanya perulangan atau struktur bertingkat (`elif`).

# KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa penggunaan statement conditional (percabangan) dalam bahasa pemrograman Python merupakan elemen esensial untuk mengontrol alur eksekusi program serta mengambil keputusan secara otomatis berdasarkan kriteria tertentu. Seluruh materi utama telah berhasil diterapkan dengan baik melalui pembuatan program interaktif yang memanfaatkan fungsi input() serta konsisten menggunakan penamaan variabel berakhiran _1003.

Penerapan tersebut mencakup enam konsep utama percabangan, yaitu:

1. Percabangan if Tunggal: Berfungsi untuk mengevaluasi satu kondisi sederhana, seperti melakukan validasi awal kuota tiket guna memastikan input pengguna bernilai valid sebelum diproses lebih lanjut.

2. Percabangan if-else: Digunakan untuk mengambil keputusan antara dua jalur alternatif, seperti mengevaluasi kelayakan penerimaan bonus souvenir berdasarkan batas minimum total pembayaran.

3. Percabangan if-elif-else: Digunakan untuk menangani kondisi majemuk yang dievaluasi secara berurutan, seperti mengevaluasi syarat kualifikasi usia dan kepemilikan SIM C pada pengunjung wahana.

4. Percabangan match-case: Berfungsi untuk melakukan pencocokan nilai diskrit (pattern matching) secara efisien dan rapi, seperti penentuan nama paket wahana beserta harga satuannya.

5. Multi-if Terpisah: Digunakan untuk mengevaluasi beberapa kondisi secara independen dan kumulatif, seperti akumulasi persentase diskon bertingkat (diskon belanja besar, status member, kode promo, dan jumlah rombongan).

6. Penggabungan Operator Logika: Berfungsi untuk mengombinasikan beberapa syarat pengecekan menggunakan operator and, !=, maupun in, sehingga alur logika program tetap terstruktur, bersih, dan terhindar dari penggunaan percabangan bersarang (nested-IF).
