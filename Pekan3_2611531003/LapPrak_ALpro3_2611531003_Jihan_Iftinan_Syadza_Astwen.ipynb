## 2.1 Dasar Teori

Operator adalah simbol khusus yang digunakan untuk melakukan operasi tertentu pada satu atau lebih nilai atau variabel (operand). Dalam bahasa pemrograman Python, terdapat berbagai jenis operator yang dikelompokkan berdasarkan fungsinya masing-masing:

1. ​Operator Aritmatika: Digunakan untuk melakukan operasi matematika dasar. Meliputi penjumlahan (+), pengurangan (-), perkalian (*), pembagian riil (/), pembagian bulat (//), sisa bagi/modulus (%), dan perpangkatan (**).

2. ​Operator Perbandingan: Digunakan untuk membandingkan dua buah nilai. Hasil dari operasi ini selalu berupa nilai boolean (True atau False). Meliputi sama dengan (==), tidak sama dengan (!=), lebih besar dari (>), lebih kecil dari (<), lebih besar atau sama dengan (>=), serta lebih kecil atau sama dengan (<=).

3. ​Operator Logika: Digunakan untuk menggabungkan beberapa kondisi atau pernyataan boolean. Meliputi and (akan bernilai benar jika semua kondisi benar), or (akan bernilai benar jika salah satu kondisi benar), dan not (kebalikan dari nilai boolean).

4. Operator Assignment (Penugasan): Digunakan untuk memberikan atau mengisi nilai ke dalam suatu variabel, yang sering kali digabungkan dengan operasi aritmatika. Contohnya seperti =, +=, -=, *=, dan /=.

5. ​Operator Keanggotaan (Membership): Digunakan untuk memeriksa apakah suatu nilai atau unsur tertentu berada di dalam sebuah wadah seperti string, list, atau tuple. Operator yang digunakan adalah in dan not in.

6. ​Operator Identitas (Identity): Digunakan untuk membandingkan lokasi memori dari dua objek variabel, bukan sekadar nilai isinya. Operator yang digunakan adalah is (bernilai benar jika kedua variabel menunjuk objek yang sama) dan is not.

7. ​Operator Bitwise: Digunakan untuk melakukan operasi pada level bit (biner) dari angka integer. Meliputi AND (&), OR (|), XOR (^), NOT (~), shift left (<<), dan shift right (>>).

## 2.2.1 Aritmatika_2611531003.py

```python
# Buat file dengan nama Aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: a_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_1003 = int(input("Input angka-1: "))
angka2_1003 = int(input("Input angka-2: "))

# Penjumlahan
hasil_penjumlahan_1003 = angka1_1003 + angka2_1003
print("\n0perator Penjumlahan")
print("Hasil =", hasil_penjumlahan_1003)

# Pengurangan
hasil_pengurangan_1003 = angka1_1003 - angka2_1003
print("\nOperator Pengurangan")
print("Hasil =", hasil_pengurangan_1003)

# Perkalian
hasil_perkalian_1003 = angka1_1003 * angka2_1003
print("\nOperator Perkalian")
print("Hasil =", hasil_perkalian_1003)

# Pembagian, Pembagian Bulat, dan sisa bagi
if angka2_1003 != 0:
    hasil_pembagian_1003 = angka1_1003 / angka2_1003
    print("\nOperator Pembagian")
    print("Hasil =", hasil_pembagian_1003)

    hasil_pembagian_bulat_1003 = angka1_1003 // angka2_1003
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_pembagian_bulat_1003)

    hasil_sisa_bagi_1003 = angka1_1003 % angka2_1003
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_sisa_bagi_1003)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_pangkat_1003 = angka1_1003 ** angka2_1003
print("\nOperator Pangkat")
print("Hasil =", hasil_pangkat_1003)
```

**Penjelasan Kode:**

1. Input dan Konversi Data (angka1_1003 = int(input(...))): Kode ini berfungsi untuk menerima masukan angka dari pengguna melalui terminal secara interaktif, kemudian langsung dikonversi menjadi tipe data integer menggunakan fungsi int() agar siap dihitung secara matematis.

2. ​Operator Penjumlahan & Pengurangan (+ dan -): Baris kode ini melakukan operasi penambahan dan pengurangan antara angka1_1003 dan angka2_1003, lalu hasilnya disimpan ke dalam variabel baru untuk dicetak ke layar.
​
3. Operator Perkalian & Pangkat (* dan **): Digunakan untuk mengalikan nilai kedua angka serta memangkatkan angka pertama (angka1_1003) dengan angka kedua (angka2_1003).
​
4. Validasi dan Pembagian (/, //, %): Blok kode if angka2_1003 != 0 digunakan sebagai pengaman agar program tidak error akibat pembagian dengan angka nol. Di dalamnya terdapat operasi pembagian biasa, pembagian bulat, serta sisa bagi (modulus).
​

**Penjelasan Output:**

1. ​Operator Penjumlahan: Menampilkan hasil total penjumlahan dari kedua angka yang dimasukkan oleh pengguna.
​
2. Operator Pengurangan: Menampilkan hasil selisih atau pengurangan antara angka pertama dan angka kedua.
​
3. Operator Perkalian: Menampilkan hasil kali dari kedua angka yang diinputkan.
​
4. Operator Pembagian (Biasa, Bulat, & Sisa Bagi): Menampilkan hasil pembagian riil, pembagian yang dibulatkan kebawah, serta sisa hasil bagi jika angka kedua valid (bukan nol). Jika bernilai nol, program akan menampilkan peringatan “Angka kedua tidak boleh bernilai 0.”
​

## 2.2.2 Perbandingan_2611531003.py

```python
# Buat file dengan nama Perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a_1234
# Progran ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_1003 = int(input("Input angka-1: "))
angka2_1003 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_lebih_besar_1003 = angka1_1003 > angka2_1003
print("\nOperator Lebih besar dari")
print("angka1 > angka2 =", hasil_lebih_besar_1003)

# Lebih kecil dari
hasil_lebih_kecil_1003 = angka1_1003 < angka2_1003
print("\nOperator Lebih kecil dari")
print("angka1 < angka2 =", hasil_lebih_kecil_1003)

# Lebih besar dari atau sama dengan
hasil_lebih_besar_sama_1003 = angka1_1003 >= angka2_1003
print("\nOperator Lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_lebih_besar_sama_1003)

# Lebih kecil dari atau sama dengan
hasil_lebih_kecil_sama_1003 = angka1_1003 <= angka2_1003
print("\nOperator Lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_lebih_kecil_sama_1003)

# Sama dengan
hasil_sama_dengan_1003 = angka1_1003 == angka2_1003
print("\nOperator Sama dengan")
print("angka1 == angka2 =", hasil_sama_dengan_1003)

# Tidak sama dengan
hasil_tidak_sama_dengan_1003 = angka1_1003 != angka2_1003
print("\nOperator Tidak sama dengan")
print("angka1 != angka2 =", hasil_tidak_sama_dengan_1003)

# Tambahan: Perbandingan berantai dalam Python
hasil_perbandingan_berantai_1003 = 0 < angka1_1003 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil_perbandingan_berantai_1003)

hasil_perbandingan_berantai_2_1003 = 0 < angka2_1003 < 100
print("0 < angka2 < 100 =", hasil_perbandingan_berantai_2_1003)
```

3. ​Perbandingan Berantai (0 < angka1_1003 < 100): Memanfaatkan fitur bawaan Python untuk memeriksa apakah suatu nilai berada di dalam rentang tertentu dalam satu baris ekspresi yang ringkas dan bersih.

2. ​Operator Lebih kecil dari (<): Menampilkan nilai True jika angka pertama lebih kecil dari angka kedua.

3. ​Operator Lebih besar / kecil dari atau sama dengan (>= & <=): Menampilkan nilai True jika kondisi relasi batas nilai tersebut terpenuhi.

## 2.2.3 Logika_2611531003.py

```python
# Buat file dengan nama Logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a_1234
# Progran ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_1003 = input("Input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_1003 = input("Input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 =", a1_1003)
print("A2 =", a2_1003)

# Konjungsi: Bernilai True jika keduanya True
hasil_konjungsi_1003 = a1_1003 and a2_1003
print("\nKonjungsi (AND)")
print("A1 AND A2 =", hasil_konjungsi_1003)

# Disjungsi: Bernilai True jika salah satu atau keduanya True
hasil_disjungsi_1003 = a1_1003 or a2_1003
print("\nDisjungsi (OR)")
print("A1 OR A2 =", hasil_disjungsi_1003)

# Negasi A1: Membalik nilai A1
hasil_negasi_1_1003 = not a1_1003
print("\nNegasi A1 (NOT)")
print("NOT A1 =", hasil_negasi_1_1003)

# Negasi A2: Membalik nilai A2
hasil_negasi_2_1003 = not a2_1003
print("\nNegasi A2 (NOT)")
print("NOT A2 =", hasil_negasi_2_1003)

# XOR: Bernilai True jika kedua nilai berbeda
hasil_XOR_1003 = a1_1003 != a2_1003
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_XOR_1003)
```

**Penjelasan Kode:**

1. ​Input Boolean Interaktif (a1_1003 & a2_1003): Mengambil input teks dari pengguna, lalu menggunakan .strip().lower() == "true" agar program tidak peka terhadap huruf besar-kecil (case-insensitive) dan langsung mengonversinya menjadi tipe data boolean (True atau False).

2. ​Operator Konjungsi / AND (and): Berfungsi untuk menggabungkan dua kondisi. Hasilnya hanya akan bernilai True jika kedua variabel bernilai benar.

3. ​Operator Disjungsi / OR (or): Menghasilkan nilai True jika salah satu atau kedua variabel bernilai benar.

4. ​Operator Negasi / NOT (not): Berfungsi untuk membalikkan nilai kebenaran suatu variabel, di mana nilai True akan menjadi False dan sebaliknya.

5. ​Operator XOR Eksklusif (!=): Memanfaatkan operator tidak sama dengan untuk mensimulasikan gerbang logika XOR, di mana hasilnya akan bernilai True jika kedua input memiliki nilai yang berbeda.


**​Penjelasan Output:**

1. ​Cetak Nilai A1 & A2: Menampilkan kembali nilai boolean akhir yang telah diproses dari input pengguna.
​
2. Konjungsi (AND): Menampilkan hasil evaluasi True atau False berdasarkan operasi AND dari kedua variabel.
​
3. Disjungsi (OR): Menampilkan hasil evaluasi operasi OR yang bernilai benar jika ada minimal satu nilai True.
​
4. Negasi A1 & A2 (NOT): Menampilkan kebalikan nilai kebenaran dari masing-masing variabel secara terpisah.
​
5. Disjungsi Eksklusif (XOR): Menampilkan status True apabila nilai A1 dan A2 tidak sama, dan False jika keduanya sama.

## 2.2.4 Assignment_2611531003.py

```python
# Buat file dengan nama Assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a_1234
# Progran ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_1003 = int(input("Input angka-1: "))
angka2_1003 = int(input("Input angka-2: "))

print("\nnilai awal angka1 =", angka1_1003)
print("nilai awal angka2 =", angka2_1003)

# Assignment biasa
hasil_assignment_biasa_1003 = angka1_1003
print("\nAssignment biasa (=)")
print("Hasil =", hasil_assignment_biasa_1003)

# Assignment penambahan
hasil_assignment_penambahan_1003 = angka1_1003
hasil_assignment_penambahan_1003 += angka2_1003
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_assignment_penambahan_1003)

# Assignment pengurangan
hasil_assignment_pengurangan_1003 = angka1_1003
hasil_assignment_pengurangan_1003 -= angka2_1003
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_assignment_pengurangan_1003)

# Assignment perkalian
hasil_assignment_perkalian_1003 = angka1_1003
hasil_assignment_perkalian_1003 *= angka2_1003
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_assignment_perkalian_1003)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_1003 != 0:
    hasil_assignment_pembagian_1003 = angka1_1003
    hasil_assignment_pembagian_1003 /= angka2_1003
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_assignment_pembagian_1003)
    # Operator tambahan
    hasil_assignment_pembagian_bulat_1003 = angka1_1003
    hasil_assignment_pembagian_bulat_1003 //= angka2_1003
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_assignment_pembagian_bulat_1003)
    hasil_assignment_sisa_bagi_1003 = angka1_1003
    hasil_assignment_sisa_bagi_1003 %= angka2_1003
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_assignment_sisa_bagi_1003)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: Assignment perpangkatan
hasil_assignment_perpangkatan_1003 = angka1_1003
hasil_assignment_perpangkatan_1003 **= angka2_1003
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_assignment_perpangkatan_1003)
```

**Penjelasan Kode:**

1. ​Input dan Konversi Data (angka1_1003 & angka2_1003): Berfungsi untuk mengambil data angka dari pengguna melalui terminal secara interaktif, lalu langsung dikonversi ke tipe data integer menggunakan fungsi int().

2. ​Assignment Biasa (=): Digunakan untuk menyalin atau memasukkan nilai dasar dari suatu ekspresi ke dalam variabel penampung.
​
3. Assignment Aritmatika (+=, -=, *=, /=): Merupakan bentuk penugasan ringkas di mana nilai variabel saat ini akan langsung dioperasikan dengan angka kedua, dan hasilnya disimpan kembali ke variabel tersebut.

4. ​Validasi dan Assignment Khusus (//=, %=, **=): Meliputi blok if-else untuk pengaman pembagian nol, serta penerapan penugasan khusus untuk pembagian bulat, sisa bagi (modulus), dan perpangkatan secara efisien dalam satu baris ekspresi.


**​Penjelasan Output:**

1. Nilai Awal Angka: Menampilkan data awal dari angka1 dan angka2 yang baru saja diinputkan oleh pengguna.

2. ​Assignment Biasa (=): Menampilkan nilai variabel hasil penugasan dasar.
​
3. Assignment Penambahan (+=) & Pengurangan (-=): Menampilkan hasil akhir variabel setelah dijumlahkan atau dikurangkan dengan nilai kedua.
​
4. Assignment Perkalian (*=): Menampilkan hasil akhir variabel setelah dikalikan dengan nilai kedua.
​
5. Assignment Pembagian, Pembagian Bulat, & Sisa Bagi (/=, //=, %=): Menampilkan hasil operasi pembagian riil, pembagian bulat, atau sisa bagi apabila angka kedua bukan nol. Jika bernilai nol, akan memunculkan pesan peringatan bahwa pembagian tidak dapat dilakukan.
​
6. Assignment Perpangkatan (**=): Menampilkan hasil akhir variabel setelah dipangkatkan dengan angka kedua.

## 2.2.5 Lainnya_2611531003.py

```python
# Buat file dengan nama Lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a_1234
# Progran ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_1003 = input("Masukkan beberapa data, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_list_1003 = [int(angka.strip()) for angka in input_data_1003.split(",")]

nilai_dicari_1003 = int(input("Masukkan nilai yang ingin dicari: "))

# Operator in
hasil_in_1003 = nilai_dicari_1003 in data_list_1003
print("\nOperator keanggotaan IN")
print(nilai_dicari_1003, "in", data_list_1003, "=", hasil_in_1003)

# Operator not in
hasil_not_in_1003 = nilai_dicari_1003 not in data_list_1003
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_1003, "not in", data_list_1003, "=", hasil_not_in_1003)


print("\n==================================")
print("2. OPERATOR IDENTITAS")
print("==================================")

# Objek1 menggunakan list dari input pengguna
objek1_1003 = data_list_1003

# Objek2 menggunakan list dari input pengguna
objek2_1003 = objek1_1003

# Objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1003 = data_list_1003.copy()

print("Objek1 =", objek1_1003)
print("Objek2 =", objek2_1003)
print("Objek3 =", objek3_1003)

# Operator is
hasil_is_1003 = objek1_1003 is objek2_1003
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_is_1003)

# Operator is not
hasil_is_not_1003 = objek1_1003 is not objek3_1003
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_is_not_1003)

# Perbandingan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_1003 is objek3_1003)
print("objek1 == objek3 =", objek1_1003 == objek3_1003)
```

**Penjelasan Kode:**

1. ​Input Data Dinamis dan List Comprehension (input_data_1003): Mengambil string masukan dari pengguna yang dipisahkan oleh koma, lalu memecahnya menggunakan .split(",") dan mengubah setiap elemennya menjadi integer secara otomatis ke dalam bentuk struktur data list.

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa penggunaan operator dalam bahasa pemrograman Python merupakan elemen esensial untuk mengolah, membandingkan, serta memanipulasi data secara dinamis. Seluruh materi utama telah berhasil diterapkan dengan baik melalui pembuatan program interaktif yang memanfaatkan fungsi input() serta konsisten menggunakan penamaan variabel berakhiran _1003.

​Penerapan tersebut mencakup tujuh jenis operator utama, yaitu:

1. ​Operator Aritmatika: Berfungsi untuk melakukan perhitungan matematika dasar seperti penjumlahan, pengurangan, perkalian, pembagian, sisa bagi, hingga perpangkatan.

2. ​Operator Perbandingan: Digunakan untuk mengevaluasi hubungan antar nilai yang secara otomatis menghasilkan nilai kebenaran boolean (True atau False).
​
3. Operator Logika: Digunakan untuk menggabungkan berbagai kondisi majemuk melalui operasi konjungsi, disjungsi, negasi, dan eksklusif.
​
4. Operator Assignment (Penugasan): Berfungsi untuk memberikan serta memperbarui nilai suatu variabel secara ringkas melalui penggabungan operasi aritmatika.
​
5. Operator Keanggotaan: Digunakan untuk memeriksa keberadaan suatu nilai tertentu di dalam struktur data menggunakan in dan not in.
​
6. Operator Identitas: Digunakan untuk membandingkan lokasi memori fisik antar objek menggunakan is dan is not guna membedakan kesamaan nilai dan alamat memori.
​
7. Operator Bitwise: Berfungsi untuk melakukan pengolahan data langsung pada tingkat bit biner, lengkap dengan operasi logika biner serta pergeseran posisi bit (shift).
